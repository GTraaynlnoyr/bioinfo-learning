```python
from Bio.Seq import Seq
my_seq = Seq("ATGCGTACGTAGCTAGCTAGC")

print("原始序列：", my_seq)
print("原始长度：", len(my_seq))
```

```python
rev_comp = my_seq.reverse_complement()
print("反向互补序列：", rev_comp)

rna = my_seq.transcribe()
print("转录为RNA：", rna)

protein = my_seq.translate()
print("翻译为蛋白质：", protein)
```
