```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'Gene': ['TP53', 'BRCA1', 'EGFR', 'MYC'],
    'Tumor': [150, 200, 320, 80],
    'Normal': [50, 40, 100, 60]
}

df = pd.DataFrame(data)
    
df['FoldChange'] = df['Tumor'] / df['Normal']

df
```
